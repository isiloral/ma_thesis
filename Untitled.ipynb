```python
import requests

api_url = "https://www.data.gv.at/katalog/api/3/action/package_show"
dataset_id = "d0145df8-7f6d-46e1-9bc6-ee7897054104"

response = requests.get(api_url, params={"id": dataset_id})
data = response.json()["result"]

# Look through the resources for a JSON or CSV
for res in data["resources"]:
    if res["format"].lower() in ["json", "csv"]:
        print("Found resource:", res["name"])
        print("Download URL:", res["url"])
```

```python
import requests
import re

url = "https://www.data.gv.at/katalog/api/3/action/package_show"
dataset_id = "d0145df8-7f6d-46e1-9bc6-ee7897054104"

response = requests.get(url, params={"id": dataset_id})
notes = response.json()["result"]["notes"]

# Search for the German word "Flächentyp" and extract 400 chars around it
keyword = "F_TYP"
start = notes.find(keyword)

if start != -1:
    snippet = notes[start:start+500]  # Grab a chunk starting from Flächentyp
    print("Found 'Flächentyp':\n")
    print(snippet)
else:
    print("'Flächentyp' not found in metadata notes.")
```
